## Convert to JSON-friendly format
This is the shape we'd save to MongoDB and serve via the Node API.

```python
import json

def df_to_statement(df, period_type):
    """Convert a yfinance statement DataFrame to a list of dicts for MongoDB storage."""
    periods = [col.strftime('%Y-%m-%d') for col in df.columns]
    rows = []
    for line_item, values in df.iterrows():
        row = {
            'label': line_item,
            'period_type': period_type,  # 'annual' or 'quarterly'
            'values': {
                col.strftime('%Y-%m-%d'): (None if pd.isna(v) else int(v))
                for col, v in values.items()
            }
        }
        rows.append(row)
    return {'periods': periods, 'rows': rows}

annual_income_json = df_to_statement(annual_income, 'annual')

print('Periods:', annual_income_json['periods'])
print('First 3 rows:')
print(json.dumps(annual_income_json['rows'][:3], indent=2))
```
